```python
import torch
from diffusers import DiffusionPipeline
from diffusers.utils import load_image

# switch to "mps" for apple devices
pipe = DiffusionPipeline.from_pretrained("Qwen/Qwen-Image-Edit-2511", dtype=torch.bfloat16, device_map="cuda")
pipe.load_lora_weights("ScottzillaSystems/qwen-image-edit-plus-nsfw-lora")
```

```python
prompt = "make him wear hat"   #@param {type:"string"}
image_path = "/content/images.jpg" #@param {type:"string"}
input_image = load_image(image_path)

image = pipe(image=input_image, prompt=prompt).images[0]
```
